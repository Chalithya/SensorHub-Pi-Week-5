```python
import requests
import sqlite3

SERVER_URL = "http://127.0.0.1:5000/api/data"
DB_FILE = "data.db"

def setup_db():
    conn = sqlite3.connect(DB_FILE)
    conn.execute('''
        CREATE TABLE IF NOT EXISTS sensor_data (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            data TEXT,
            timestamp DATETIME DEFAULT CURRENT_TIMESTAMP
        )
    ''')
    conn.commit()
    conn.close()

def collect_data():
    try:
        response = requests.get(SERVER_URL, timeout=5)
        if response.status_code == 200:
            data = response.text
            conn = sqlite3.connect(DB_FILE)
            conn.execute("INSERT INTO sensor_data (data) VALUES (?)", (data,))
            conn.commit()
            conn.close()
            print("Data collected and stored.")
        else:
            print(f"Server error: {response.status_code}")
    except Exception as e:
        print("Connection failed:", e)

# Step 3: Run once
setup_db()
collect_data()
```

```python
# See DB result
conn = sqlite3.connect("data.db")
rows = conn.execute("SELECT * FROM sensor_data").fetchall()
for row in rows:
    print(row)
conn.close()
```

```python
import requests
import sqlite3
import time

SERVER_URL =  "http://127.0.0.1:5000/api/data"  
DB_FILE = "data.db"

def setup_db():
    conn = sqlite3.connect(DB_FILE)
    conn.execute('''
        CREATE TABLE IF NOT EXISTS sensor_data (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            data TEXT,
            timestamp DATETIME DEFAULT CURRENT_TIMESTAMP
        )
    ''')
    conn.commit()
    conn.close()

def collect_data():
    try:
        response = requests.get(SERVER_URL, timeout=5)
        if response.status_code == 200:
            data = response.text.strip()  
            conn = sqlite3.connect(DB_FILE)
            conn.execute("INSERT INTO sensor_data (data) VALUES (?)", (data,))
            conn.commit()
            conn.close()
            print("Data collected and stored.")
        else:
            print(f"Server error: {response.status_code}")
    except Exception as e:
        print("Connection failed:", e)

def main():
    print("Starting data collector...")
    setup_db()
    try:
        while True:
            collect_data()
            time.sleep(30)  # collect data every 30 seconds
    except KeyboardInterrupt:
        print("\n Stopped by user.")

if __name__ == "__main__":
    main()  
```
